```python
import pandas as pd

df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date'].dtype
```

```python
df['Year'] = df['Date'].dt.year
```

```python
df.groupby(['Year','Card Type']).agg(
    total_spend=('Amount', 'sum')
).reset_index()
```

```python
df.groupby('Year')['Amount'].sum().reset_index()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Aggregate
pivot = (
    df
    .groupby('Exp Type')['Amount']
    .sum()
    .unstack(fill_value=0)
)
pivot
```

```python
ax = pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Spend by Card Type and Expense Type')
plt.xlabel('Card Type')
plt.ylabel('Total Amount')
plt.legend(title='Expense Type', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```
